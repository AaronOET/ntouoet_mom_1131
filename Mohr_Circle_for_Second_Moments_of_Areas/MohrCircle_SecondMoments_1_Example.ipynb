Section 10-3.1, Figs. 10-22 to 10-25, and Example Problem 10-13.
The left panel shows the area with its axes; the right panel shows Mohr's circle. Each axis on the area is tagged with its point on the circle:

| Point | Coordinates | Axis on the area |
| --- | --- | --- |
| A | $(I_x,\ I_{xy})$ | $x$ |
| B | $(I_y,\ -I_{xy})$ | $y$ |
| C | $\left(\frac{I_x+I_y}{2},\ 0\right)$ | centre |
| D | $(I_{max},\ 0)$ | $u$ (principal, maximum) |
| E | $(I_{min},\ 0)$ | $v$ (principal, minimum) |
| F | $(I_{x'},\ I_{x'y'})$ | $x'$, rotated $\theta$ counterclockwise from $x$ |
| G | $(I_{y'},\ -I_{x'y'})$ | $y'$ |

$$
OC = \frac{I_x+I_y}{2},\qquad R = \left[\left(\frac{I_x-I_y}{2}\right)^2 + I_{xy}^2\right]^{1/2},\qquad I_{max,\,min} = OC \pm R
$$

**Sign convention (as in the text):** $I_{x'}$ is plotted to the right; $I_{x'y'}$ is plotted **upward** when positive.
Rotating the axes by $\theta$ counterclockwise moves F by $2\theta$ counterclockwise: **all angles on Mohr's circle are double those on the area.**

## Build the circle: Example Problem 10-13

The sliders start at the values of Example 10-13 (unequal-leg angle, axes through the centroid), in $10^3$ mm$^4$:

$$I_x = 808,\qquad I_y = 388,\qquad I_{xy} = -323$$

Things to do:

1. **Construction** dropdown, steps 1 → 4: plot A (808, −323), then B (388, 323), then join AB to find C, then draw the circle (Fig. 10-25a, b).
   Read off $OC = 598$, $R = 385$, $I_{max} = 983$, $I_{min} = 213$.
2. **θ slider**: rotate the $x'$-axis and watch F move twice as fast around the circle. At **θ = 28.5°** F reaches D and G reaches E.
   Then $I_{x'y'} = 0$ and $x'$ is the principal $u$-axis, 28.5° counterclockwise from $x$ (Fig. 10-25c, d).
   A is below the horizontal axis here because $I_{xy} < 0$, so D is counterclockwise from A.
3. At **θ = 90°**, F lands on B: the $x'$-axis is now the old $y$-axis.
4. Change $I_x$, $I_y$, $I_{xy}$:
   * $I_{xy} > 0$ moves A above the axis (Fig. 10-22) and the $u$-axis turns clockwise from $x$.
   * $I_x < I_y$ puts A left of C.
   * $I_x = I_y$ and $I_{xy} = 0$ shrink the circle to a point, so every axis is principal.
   * $I_{xy}^2 > I_x I_y$ pushes the circle left of O. No real area can do this, and a warning appears.
