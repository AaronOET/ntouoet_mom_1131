The rectangles must not overlap. The module finds the centroid and $I_x$, $I_y$, $I_{xy}$ about centroidal axes
(parallel-axis theorem), draws the section with Mohr's circle, and gives a $\theta$ slider for the rotated axes $x'$, $y'$.

Edit the name and the rectangles below, then run the cell. The example is a C-section with a 20 × 160 mm web,
a 100 mm top flange and a 50 mm bottom flange, all 20 mm thick.

```python
my_section = mc.Section("C-section 160 x 100/50 x 20",
                        [(0, 0, 20, 160),      # web
                         (20, 140, 80, 20),    # top flange
                         (20, 0, 30, 20)])     # bottom flange                         

mc.custom_section_app(my_section)
```
