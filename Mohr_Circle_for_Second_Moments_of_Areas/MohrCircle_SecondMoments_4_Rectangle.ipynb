## Rectangle $b \times h$ about its centroid

For a rectangle of width $b$ (along $x$) and height $h$ (along $y$), with axes through the centroid:

$$
I_x = \frac{bh^3}{12}, \qquad I_y = \frac{hb^3}{12}, \qquad I_{xy} = 0
$$

$I_{xy} = 0$ because $x$ and $y$ are both axes of symmetry. So A $= (I_x, 0)$ and B $= (I_y, 0)$ lie **on** the horizontal axis and AB is the horizontal diameter:

$$
OC = \frac{I_x + I_y}{2} = \frac{bh\,(b^2 + h^2)}{24}, \qquad
R = \frac{|I_x - I_y|}{2} = \frac{bh\,|h^2 - b^2|}{24}
$$

The $x$ and $y$ axes are always the principal axes, and only $b$ and $h$ decide which one is the maximum.

| Shape | A, B on the circle | Principal axes |
| --- | --- | --- |
| Tall, $h > b$ | A = D (right end), B = E | $u = x$, $v = y$ |
| Wide, $b > h$ | A = E (left end), B = D | $u = y$, $v = x$ |
| Square, $b = h$ | $R = 0$: the circle is a single point | every axis |

For rotated axes, F moves $2\theta$ counterclockwise from A:

$$
I_{x'} = OC + (I_x - I_y)/2 \cdot \cos 2\theta, \qquad I_{x'y'} = (I_x - I_y)/2 \cdot \sin 2\theta
$$

and $|I_{x'y'}|$ is largest, equal to $R$, at $\theta = 45^\circ$.

Things to try:

2. **Make it wide** (b > h). A moves to the left end of the diameter and the $u$-axis switches to $y$.
3. **Make it square** (b = h). The circle shrinks to a point and every axis is principal.
4. **Double h** with b fixed. $I_x$ grows 8 times ($h^3$) but $I_y$ only 2 times. This is why beams are made deep rather than wide.
5. **θ = 45°**: F is at the top or bottom of the circle, so $|I_{x'y'}| = R$ is at its largest, and $I_{x'} = I_{y'} = OC$.
