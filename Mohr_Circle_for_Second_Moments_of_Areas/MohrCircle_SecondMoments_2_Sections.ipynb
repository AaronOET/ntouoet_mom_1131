## Example: rectangle 60 × 120 mm

A rectangle of width $b = 60$ mm and height $h = 120$ mm, axes through the centroid. Units: $10^3$ mm$^4$.

**Second moments.** The $x$ and $y$ axes are both axes of symmetry, so $I_{xy} = 0$:

$$
I_x = \frac{bh^3}{12} = \frac{60(120)^3}{12} = 8640, \qquad
I_y = \frac{hb^3}{12} = \frac{120(60)^3}{12} = 2160, \qquad
I_{xy} = 0
$$

**Mohr's circle.** A $= (8640,\ 0)$ and B $= (2160,\ 0)$ both lie **on** the horizontal axis, so AB is the horizontal diameter:

$$
OC = \frac{8640 + 2160}{2} = 5400, \qquad R = \sqrt{\left(\frac{8640 - 2160}{2}\right)^2 + 0^2} = 3240
$$

$$
I_{max} = OC + R = 8640 = I_x \quad (\text{A} = \text{D}), \qquad I_{min} = OC - R = 2160 = I_y \quad (\text{B} = \text{E})
$$

**Principal axes.** $2\theta_p = 0$: the $x$-axis is already the $u$-axis and the $y$-axis is the $v$-axis. Any axis of symmetry is a principal axis.

**Rotated axes.** For $\theta = 30^\circ$, F is $2\theta = 60^\circ$ counterclockwise from A:

$$
I_{x'} = OC + R\cos 60^\circ = 5400 + 3240(0.5) = 7020, \qquad
I_{x'y'} = R\sin 60^\circ = 3240\sin 60^\circ = 2805.9
$$

Drag θ in the app below. At $\theta = 30^\circ$ you should read F $= (7020,\ 2805.92)$, and at $\theta = 90^\circ$ F reaches B: the $x'$-axis is then the weak $y$-axis, with $I_{x'} = I_{min} = 2160$.
